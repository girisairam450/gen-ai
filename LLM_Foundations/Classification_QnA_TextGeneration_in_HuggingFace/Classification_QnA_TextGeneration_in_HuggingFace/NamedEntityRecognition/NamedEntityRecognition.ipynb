```python
from transformers import pipeline
```

https://huggingface.co/dslim/bert-base-NER

- O	Outside of a named entity
- B-MISC	Beginning of a miscellaneous entity right after another miscellaneous entity
- I-MISC	Miscellaneous entity
- B-PER	Beginning of a person’s name right after another person’s name
- I-PER	Person’s name
- B-ORG	Beginning of an organization right after another organization
- I-ORG	organization
- B-LOC	Beginning of a location right after another location
- I-LOC	Location

```python
ner_pipeline = pipeline("ner", model="dslim/bert-large-NER")

ner_pipeline
```

```python
text = "Apple Inc. is set to launch its latest iPhone model next month."

print(text)

ner_pipeline(text)
```

aggregation_strategy is used to group the entities in the predictions. It will group "Apple Inc" together

```python
text = "Apple Inc. is set to launch its latest iPhone model next month."

print(text)

ner_pipeline(text, aggregation_strategy="simple")
```

```python
texts = [
    "The Nile River is the longest river in Africa, flowing through multiple countries including Egypt and Sudan.",
    "The Statue of Liberty stands in New York Harbor."
]

print(texts)

ner_pipeline(texts, aggregation_strategy="simple")
```

```python
ner_results = ner_pipeline(texts, aggregation_strategy="simple")

for idx, (text, result) in enumerate(zip(texts, ner_results), 1):
    print(f"\nText {idx}: {text}")

    if result and isinstance(result[0], dict):
        print("Entities:")
        for entity in result:
            for key, value in entity.items():
                print(f"  - {key}: {value}")
```

```python
import tensorflow as tf

from transformers import TFBertForTokenClassification, BertTokenizer
```

```python
DBMDZ_MODEL = "dbmdz/bert-large-cased-finetuned-conll03-english"

DBMDZ_MODEL
```

```python
tokenizer = BertTokenizer.from_pretrained(DBMDZ_MODEL)

tokenizer
```

```python
model = TFBertForTokenClassification.from_pretrained(
    DBMDZ_MODEL, from_pt=True)

model
```

```python
text = "Apple Inc. is a technology company based in Cupertino, California. It was founded by Steve Jobs, Steve Wozniak, and Ronald Wayne."

text
```

```python
tokens = tokenizer.encode_plus(text, return_tensors="tf")

tokens
```

```python
outputs = model(tokens)

outputs
```

##### obtain predicted labels by finding the index with the highest probability

```python
predicted_labels = tf.argmax(outputs.logits, axis=2).numpy()

predicted_labels
```

##### converting back

```python
predicted_tokens = tokenizer.convert_ids_to_tokens(tokens["input_ids"][0])

predicted_tokens
```

```python
for token, label in zip(predicted_tokens, predicted_labels[0]):
    print(f"{token} - {model.config.id2label[label]}")
```

##### removing all the unnessary ##, O and parsing it properly for better understanding

```python
result_dict = {}

current_entity = ""

current_label = None

for token, label_id in zip(predicted_tokens, predicted_labels[0]):
    label = model.config.id2label[label_id]

    entity_type = label.split('-', 1)[1] if '-' in label else 'O'

    if entity_type != 'O':
        if token.startswith("##"):
            current_entity += token[2:]
        else:
            current_entity += " " + token if current_entity else token
        current_label = entity_type
    elif current_entity:
        result_dict[current_entity] = current_label
        current_entity = ""
        current_label = None
```

```python
for entity, label in result_dict.items():
    print(f"{entity} - {label}")
```
