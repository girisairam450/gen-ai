Using the tokenizers and the model

```python
from transformers import AutoTokenizer, TFAutoModelForQuestionAnswering

DISTILBERT_MODEL = "distilbert-base-cased-distilled-squad"

DISTILBERT_MODEL
```

```python
tokenizer = AutoTokenizer.from_pretrained(DISTILBERT_MODEL)

tokenizer
```

```python
model = TFAutoModelForQuestionAnswering.from_pretrained(DISTILBERT_MODEL)

model
```

```python
question = "Which celestial body was the focus of NASA's recent mission involving the Curiosity rover?"

question
```

##### initialized tokenizer to convert the provided question and context into a format suitable for the model

```python
inputs = tokenizer(question, context, return_tensors="tf")

inputs
```

```python
outputs = model(**inputs)

outputs
```

##### extracts the start logits from the model outputs which represent the model prediction for the starting position of the answer in the provided context

```python
start_logits = outputs.start_logits

start_logits
```

##### end logits from the model outputs representing the model prediction for the ending position of the answer in the given context

```python
end_logits = outputs.end_logits

end_logits
```

```python
print(start_logits.shape, end_logits.shape)
```

```python
import numpy as np

np.argmax(start_logits[0])
```

```python
import numpy as np

np.argmax(end_logits[0])
```

##### Mask everything apart from the tokens of the context, Unmask the [CLS] token

To convert those logits into probabilities, we will apply a softmax function — but before that, we need to make sure we mask the indices that are not part of the context. Our input is [CLS] question [SEP] context [SEP], so we need to mask the tokens of the question as well as the [SEP] token. We’ll keep the [CLS] token, however, as some models use it to indicate that the answer is not in the context.

```python
import tensorflow as tf

sequence_ids = inputs.sequence_ids()

mask = [i != 1 for i in sequence_ids]

mask[0] = False
mask = tf.constant(mask)[None]

mask
```

```python
start_logits = tf.where(mask, -10000, start_logits)

end_logits = tf.where(mask, -10000, end_logits)
```

```python
start_logits
```

```python
start_probabilities = tf.math.softmax(start_logits, axis=-1)[0].numpy()

end_probabilities = tf.math.softmax(end_logits, axis=-1)[0].numpy()
```

```python
start_probabilities
```

```python
end_probabilities
```

At this stage, we could take the argmax of the start and end probabilities — but we might end up with a start index that is greater than the end index, so we need to take a few more precautions. We will compute the probabilities of each possible start_index and end_index where start_index <= end_index, then take the tuple (start_index, end_index) with the highest probability.


compute the scores for all possible combinations of start and end positions by multiplying the softmax probabilities scores are the different answer spans in the given context

```python
start_probabilities.shape, start_probabilities[:, None].shape
```

```python
scores = start_probabilities[:, None] * end_probabilities[None, :]

scores
```

```python
scores.shape
```

##### keeps only the scores for positions where the start position is before or at the same position as the end position

Then we’ll mask the values where start_index > end_index by setting them to 0 (the other probabilities are all positive numbers). The np.triu() function returns the upper triangular part of the 2D tensor passed as an argument, so it will do that masking for us:

```python
import numpy as np

scores = np.triu(scores)

scores
```

```python
max_index = scores.argmax().item()

max_index
```

Note that the score matches with the score we got when we worked directly with the pipeline

```python
max_index = scores.argmax().item()

start_index = max_index // scores.shape[1]
end_index = max_index % scores.shape[1]

print(scores[start_index, end_index])
```
