#### Top-k

```python
text = "Once upon a time"

print(text)

answer = tg_pipeline(
    text,
    pad_token_id=tg_pipeline.tokenizer.eos_token_id,
    max_new_tokens=100,
    do_sample=True,
    top_k=3
)


print(answer[0]['generated_text'])
```

* top_k=0: This means all vocabulary tokens are considered during sampling, making it more random

* top_k=3: Only the top 3 vocabulary tokens by probability are considered during sampling, making it more constrained and focused

* top_k=25: A larger value (25 in this case) means a broader set of top probability tokens are considered during sampling, providing more diversity compared to top_k=3 but less randomness compared to top_k=0

```python
text = "Once upon a time"

print(text)

answer = tg_pipeline(
    text,
    pad_token_id=tg_pipeline.tokenizer.eos_token_id,
    max_new_tokens=100,
    do_sample=True,
    top_k=25
)

print(answer[0]['generated_text'])
```

##### Nulceus sampling

```python
text = "Once upon a time"

print(text)

answer = tg_pipeline(
    text,
    pad_token_id=tg_pipeline.tokenizer.eos_token_id,
    max_new_tokens=100,
    do_sample=True,
    top_p=0.88
)

print(answer[0]['generated_text'])
```

```python
text = "Once upon a time"

print(text)

answer = tg_pipeline(
    text,
    pad_token_id=tg_pipeline.tokenizer.eos_token_id,
    max_new_tokens=100,
    do_sample=True,
    top_p = 0.5
)

print(answer[0]['generated_text'])
```

* top_p=0.88: This means that during sampling, the algorithm will consider tokens until the cumulative probability reaches 88%. It allows for more diverse and open-ended text generation, as it includes a broader range of likely tokens.

* top_p=0.5: In this case, the algorithm will consider tokens until the cumulative probability reaches 50%. This makes the sampling more focused on a narrower set of highly probable tokens, potentially resulting in more constrained and deterministic text.

```python
text = "Once upon a time"

print(text)

answer = tg_pipeline(
    text,
    pad_token_id=tg_pipeline.tokenizer.eos_token_id,
    max_new_tokens=100,
    do_sample=True,
    top_k = 30,
    top_p = 0.8
)

print(answer[0]['generated_text'])
```

```python
import tensorflow as tf

from transformers import TFGPT2LMHeadModel, GPT2Tokenizer
```

```python
tokenizer = GPT2Tokenizer.from_pretrained('gpt2')

tokenizer
```

```python
model = TFGPT2LMHeadModel.from_pretrained('gpt2')

model
```

```python
text = "In the realm of technology, where innovation reigns supreme,"

print(text)
```

##### convert the input text into numerical token id and it returns a tensor containing the encoded sequence

```python
input_ids = tokenizer.encode(text, return_tensors="tf")

input_ids
```

```python
output = model.generate(
    input_ids,
    max_new_tokens=100,
    num_return_sequences=2,
    no_repeat_ngram_size=2,
    do_sample=True,
    top_k=50,
    top_p=0.92,
    pad_token_id=tg_pipeline.tokenizer.eos_token_id,
)

output
```

```python
generated_text = tokenizer.decode(output[0], skip_special_tokens=True)

print(generated_text)
```

```python
text = "In the realm of technology, where innovation reigns supreme,"

input_ids = tokenizer.encode(text, return_tensors="tf")

output = model.generate(
    input_ids,
    max_new_tokens=100,
    num_beams=20,
    no_repeat_ngram_size=1,
    pad_token_id=tg_pipeline.tokenizer.eos_token_id,
)

generated_text = tokenizer.decode(output[0], skip_special_tokens=True)

print(generated_text)
```
